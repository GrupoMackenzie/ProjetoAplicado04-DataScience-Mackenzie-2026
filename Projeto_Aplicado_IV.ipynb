# **Referencial Teórico**

&emsp;A modelagem de demanda no transporte aéreo constitui um clássico problema de análise de séries temporais, no qual fatores econômicos, comportamentais e sazonais convergem para determinar fluxos de tráfego (ICAO, 2006). A literatura científica divide as abordagens de previsão em três grupos metodológicos principais: modelos estatísticos econométricos, modelos aditivos modulares e algoritmos de aprendizado de máquina supervisionado.

##**Modelos Estatísticos Clássicos (ARIMA / SARIMA)**
&emsp;Os modelos da metodologia Box-Jenkins, especialmente o SARIMA (Autoregressivo Integrado de Médias Móveis com Sazonalidade), fundamentam-se na dependência linear entre observações passadas defasadas e choques estocásticos (BOX et al., 2015).
* Vantagens: Alta interpretabilidade estatística, fundamentação teórica sólida e excelente precisão para horizontes de curto prazo em séries estacionárias com sazonalidade regular.
* Limitações: Exigem estacionariedade estrita (média e variância constantes no tempo) e assumem relações estritamente lineares, apresentando fraco desempenho diante de quebras estruturais abruptas ou choques macroeconômicos (HYNDMAN; ATHANASOPOULOS, 2021).

##**Modelos Aditivos Decomponíveis (Prophet)**
&emsp;Desenvolvido para lidar com séries temporais de negócios, o Prophet modela a série através de uma formulação aditiva composta por tendência não linear, componentes sazonais múltiplos (Fourier) e impactos de feriados e eventos extraordinários.
* Vantagens: Ajuste automático a mudanças bruscas de tendência (changepoints), robustez a dados faltantes e facilidade de inclusão de eventos de calendário.
* Limitações: Menor sensibilidade a variações de curtíssimo prazo e dependência de calibração empírica dos parâmetros de flexibilidade da tendência.

##**Aprendizado de Máquina e Redes Neurais (XGBoost / LSTM)**
&emsp;Algoritmos baseados em árvores de decisão aumentadas por gradiente (Gradient Boosting como XGBoost) e redes neurais recorrentes (Long Short-Term Memory - LSTM) tratam a previsão temporal por meio de regressão supervisionada não linear e mapeamento sequencial de estados.
* Vantagens: Capacidade de capturar padrões não lineares complexos, interações entre múltiplos fatores exógenos simultâneos e ausência da premissa de normalidade residual.
* Limitações: Risco elevado de sobreajuste (overfitting), necessidade de engenharia de variáveis defasadas (lag features) e maior opacidade interpretativa (black-box).

##**Conceitos Fundamentais e Validação**
* Estacionariedade: Propriedade em que média, variância e autocorrelação são constantes ao longo do tempo. Será avaliada via teste de Dickey-Fuller Aumentado (ADF).
* Decomposição Temporal: Separação analítica da série nas componentes de Tendência (T), Sazonalidade (S) e Resíduo aleatório (I).
* Validação Temporal Cruzada (Time Series Split): Esquema de divisão que preserva a ordem cronológica, simulando o ambiente real de previsão e impedindo o vazamento de informações futuras (data leakage).
* Métricas de Erro: O desempenho preditivo será quantificado por Erro Médio Absoluto (MAE), Raiz do Erro Quadrático Médio (RMSE) e Erro Percentual Absoluto Médio (MAPE).

````python
# 1. Cria a coluna de Data (Datetime) a partir de ANO e MES
dataset['DATA'] = pd.to_datetime(dataset['ANO'].astype(str) + '-' + dataset['MES'].astype(str) + '-01')

# 2. Aplica o filtro operacional: Apenas Voos Regulares (removendo traslados e voos de treinamento)
df_regular = dataset[dataset['GRUPO_DE_VOO'] == 'REGULAR'].copy()

# 3. Agrega os dados mensalmente para a estrutura de Série Temporal
cols_to_agg = ['PASSAGEIROS_PAGOS', 'RPK', 'ASK', 'DECOLAGENS', 'ASSENTOS']
df_mensal = df_regular.groupby('DATA')[cols_to_agg].sum().reset_index()

# 4. Tratamento da Anomalia da COVID-19 (2020-2021)

# Abordagem A: Variável Dummy (Excelente para modelos SARIMAX e XGBoost)
# Definimos o período de maior impacto da pandemia com 1, e o resto com 0
df_mensal['COVID_DUMMY'] = df_mensal['DATA'].apply(
    lambda x: 1 if pd.to_datetime('2020-03-01') <= x <= pd.to_datetime('2021-12-01') else 0
)

# Abordagem B: Suavização/Interpolação (Para analisar a tendência de longo prazo sem o choque)
# Substituindo os valores do pico inicial da pandemia (mar/2020 a dez/2020) por NaN e interpolando linearmente
df_mensal['PASSAGEIROS_PAGOS_TRATADO'] = df_mensal['PASSAGEIROS_PAGOS']
mascara_covid_pico = (df_mensal['DATA'] >= '2020-03-01') & (df_mensal['DATA'] <= '2020-12-01')

# Aplica NaN no período crítico e interpola
df_mensal.loc[mascara_covid_pico, 'PASSAGEIROS_PAGOS_TRATADO'] = pd.NA
df_mensal['PASSAGEIROS_PAGOS_TRATADO'] = df_mensal['PASSAGEIROS_PAGOS_TRATADO'].astype(float).interpolate(method='linear')

# 5. Visualizando o impacto do tratamento
plt.figure(figsize=(14, 6))
plt.plot(df_mensal['DATA'], df_mensal['PASSAGEIROS_PAGOS'], label='Original (Com Choque COVID)', color='red', alpha=0.6, linewidth=2)
plt.plot(df_mensal['DATA'], df_mensal['PASSAGEIROS_PAGOS_TRATADO'], label='Tratado (Interpolado)', color='teal', linestyle='--', linewidth=2)
plt.title('Tratamento da Anomalia COVID-19 - Evolução Mensal de Passageiros Pagos', fontsize=14)
plt.xlabel('Data', fontsize=12)
plt.ylabel('Passageiros Pagos (Dezenas de Milhões)', fontsize=12)
plt.axvspan(pd.to_datetime('2020-03-01'), pd.to_datetime('2021-12-01'), color='gray', alpha=0.2, label='Período COVID (Dummy)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
````

# **Modelos**

Descrição metodológica para a criação, refinamento e validação dos modelos propostos.

````python
print("Preparando os dados para o modelo SARIMAX...")

# Preparar o dataset (Garantir que 'DATA' seja o índice com frequência mensal)
df_modelo = df_mensal.copy()
df_modelo.set_index('DATA', inplace=True)
df_modelo.index.freq = 'MS' # Frequência Mensal (Month Start)

# Definir variável alvo e variável exógena
target = 'PASSAGEIROS_PAGOS'
exog_var = 'COVID_DUMMY'

# Divisão de Treino e Teste (Deixando os últimos 18 meses para teste como baseline)
test_size = 18
train = df_modelo.iloc[:-test_size]
test = df_modelo.iloc[-test_size:]

print(f"Tamanho do Treino: {len(train)} meses")
print(f"Tamanho do Teste: {len(test)} meses\n")

# Construção do Modelo SARIMAX (Baseline com hiperparâmetros arbitrários iniciais)
# order=(p,d,q) e seasonal_order=(P,D,Q,s) onde s=12 para sazonalidade anual
modelo_base = SARIMAX(
    train[target],
    exog=train[exog_var],
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)

print("Treinando o modelo base...")
resultado_base = modelo_base.fit(disp=False)

# Previsões no conjunto de teste
previsoes = resultado_base.predict(
    start=test.index[0],
    end=test.index[-1],
    exog=test[[exog_var]]
)

# Avaliação dos Resultados (Métricas de Erro)
mae = mean_absolute_error(test[target], previsoes)
rmse = np.sqrt(mean_squared_error(test[target], previsoes))
mape = mean_absolute_percentage_error(test[target], previsoes)

print("=== Resultados do Modelo Base (SARIMAX) ===")
print(f"MAE (Erro Médio Absoluto): {mae:,.2f}")
print(f"RMSE (Raiz do Erro Quadrático Médio): {rmse:,.2f}")
print(f"MAPE (Erro Percentual Absoluto Médio): {mape*100:.2f}%\n")

# Visualização dos Resultados
plt.figure(figsize=(14, 6))
plt.plot(train.index, train[target], label='Treino', color='teal')
plt.plot(test.index, test[target], label='Teste (Real)', color='orange')
plt.plot(test.index, previsoes, label='Previsão (Baseline)', color='red', linestyle='--')
plt.title('Baseline SARIMAX: Previsão de Passageiros Pagos', fontsize=14)
plt.xlabel('Data', fontsize=12)
plt.ylabel('Passageiros Pagos', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# Exibir sumário estatístico básico do modelo
print(resultado_base.summary().tables[1])
````

# **Resultados**

Apresentar os resultados obtidos, métricas, tabelas comparativas, etc.

````python
# Resultados - Comparação com Modelo Naive (Ingênuo)

# 1. Criar um modelo Naive Sazonal (prever o valor do mesmo mês do ano anterior)
# Como o test_size é 18, pegamos os valores defasados de 12 meses do próprio df_modelo
previsoes_naive = df_modelo[target].shift(12).loc[test.index]

# 2. Calcular métricas para o Naive
mae_naive = mean_absolute_error(test[target], previsoes_naive)
rmse_naive = np.sqrt(mean_squared_error(test[target], previsoes_naive))
mape_naive = mean_absolute_percentage_error(test[target], previsoes_naive)

# 3. Consolidar os resultados em um DataFrame comparativo
df_comparacao = pd.DataFrame({
    'Modelo': ['Baseline (SARIMAX)', 'Naive Sazonal (Ano Anterior)'],
    'MAE': [mae, mae_naive],
    'RMSE': [rmse, rmse_naive],
    'MAPE (%)': [mape * 100, mape_naive * 100]
})

# Formatação para melhor visualização
df_comparacao_styled = df_comparacao.copy()
df_comparacao_styled['MAE'] = df_comparacao_styled['MAE'].map('{:,.2f}'.format)
df_comparacao_styled['RMSE'] = df_comparacao_styled['RMSE'].map('{:,.2f}'.format)
df_comparacao_styled['MAPE (%)'] = df_comparacao_styled['MAPE (%)'].map('{:.2f}%'.format)

print("=== Tabela Comparativa de Resultados ===")
display(df_comparacao_styled)

# 4. Visualização comparativa no período de teste
plt.figure(figsize=(14, 6))
plt.plot(test.index, test[target], label='Real (Teste)', color='orange', linewidth=2)
plt.plot(test.index, previsoes, label='SARIMAX (Baseline)', color='red', linestyle='--')
plt.plot(test.index, previsoes_naive, label='Naive Sazonal', color='gray', linestyle=':')
plt.title('Comparação: Real vs Previsões (SARIMAX x Naive Sazonal)', fontsize=14)
plt.xlabel('Data')
plt.ylabel('Passageiros Pagos')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
````
