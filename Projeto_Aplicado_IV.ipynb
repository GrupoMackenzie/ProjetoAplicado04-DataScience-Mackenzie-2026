```python
# Importaões e Instalações
import pandas as pd
```

```python
dados = ""
# Lendo o CSV pulando a primeira linha de metadados e forçando a leitura correta dos tipos
dataset = pd.read_csv(dados, sep=";", skiprows=1, low_memory=False)
```

```python
# 1. Visualizando o tamanho real do dataset (linhas, colunas)
print(f"Dimensões reais do dataset: {dataset.shape[0]} linhas e {dataset.shape[1]} colunas.\n")

# 2. Informações gerais sobre os tipos de dados reais de cada coluna
dataset.info()
```

```python
# 3. Verificando a quantidade de valores nulos por coluna real
valores_nulos = dataset.isnull().sum()
porcentagem_nulos = (valores_nulos / len(dataset)) * 100

analise_nulos = pd.DataFrame({
    'Nulos Total': valores_nulos,
    'Porcentagem (%)': porcentagem_nulos
})

print("Análise de valores nulos nas colunas (ordenadas de forma decrescente):")
display(analise_nulos.sort_values(by='Nulos Total', ascending=False))
```
