# **Referencial Teórico**

&emsp;A modelagem de demanda no transporte aéreo constitui um clássico problema de análise de séries temporais, no qual fatores econômicos, comportamentais e sazonais convergem para determinar fluxos de tráfego (ICAO, 2006). A literatura científica divide as abordagens de previsão em três grupos metodológicos principais: modelos estatísticos econométricos, modelos aditivos modulares e algoritmos de aprendizado de máquina supervisionado.

##**Modelos Estatísticos Clássicos (ARIMA / SARIMA)**
&emsp;Os modelos da metodologia Box-Jenkins, especialmente o SARIMA (Autoregressivo Integrado de Médias Móveis com Sazonalidade), fundamentam-se na dependência linear entre observações passadas defasadas e choques estocásticos (BOX et al., 2015).
* Vantagens: Alta interpretabilidade estatística, fundamentação teórica sólida e excelente precisão para horizontes de curto prazo em séries estacionárias com sazonalidade regular.
* Limitações: Exigem estacionariedade estrita (média e variância constantes no tempo) e assumem relações estritamente lineares, apresentando fraco desempenho diante de quebras estruturais abruptas ou choques macroeconômicos (HYNDMAN; ATHANASOPOULOS, 2021).

##**Modelos Aditivos Decomponíveis (Prophet)**
&emsp;Desenvolvido para lidar com séries temporais de negócios, o Prophet modela a série através de uma formulação aditiva composta por tendência não linear, componentes sazonais múltiplos (Fourier) e impactos de feriados e eventos extraordinários.
* Vantagens: Ajuste automático a mudanças bruscas de tendência (changepoints), robustez a dados faltantes e facilidade de inclusão de eventos de calendário.
* Limitações: Menor sensibilidade a variações de curtíssimo prazo e dependência de calibração empírica dos parâmetros de flexibilidade da tendência.

##**Aprendizado de Máquina e Redes Neurais (XGBoost / LSTM)**
&emsp;Algoritmos baseados em árvores de decisão aumentadas por gradiente (Gradient Boosting como XGBoost) e redes neurais recorrentes (Long Short-Term Memory - LSTM) tratam a previsão temporal por meio de regressão supervisionada não linear e mapeamento sequencial de estados.
* Vantagens: Capacidade de capturar padrões não lineares complexos, interações entre múltiplos fatores exógenos simultâneos e ausência da premissa de normalidade residual.
* Limitações: Risco elevado de sobreajuste (overfitting), necessidade de engenharia de variáveis defasadas (lag features) e maior opacidade interpretativa (black-box).

##**Conceitos Fundamentais e Validação**
* Estacionariedade: Propriedade em que média, variância e autocorrelação são constantes ao longo do tempo. Será avaliada via teste de Dickey-Fuller Aumentado (ADF).
* Decomposição Temporal: Separação analítica da série nas componentes de Tendência (T), Sazonalidade (S) e Resíduo aleatório (I).
* Validação Temporal Cruzada (Time Series Split): Esquema de divisão que preserva a ordem cronológica, simulando o ambiente real de previsão e impedindo o vazamento de informações futuras (data leakage).
* Métricas de Erro: O desempenho preditivo será quantificado por Erro Médio Absoluto (MAE), Raiz do Erro Quadrático Médio (RMSE) e Erro Percentual Absoluto Médio (MAPE).

## **Trabalhos Correlatos**
&emsp;A literatura recente tem demonstrado o valor da comparação entre métodos estatísticos e computacionais para o setor aéreo. Em um estudo de caso aplicado à demanda de passageiros aeroportuários, Hassani et al. (2019) utilizaram algoritmos de Machine Learning em contraponto ao clássico modelo SARIMA. Os autores demonstraram que abordagens baseadas em redes neurais e regressões avançadas conseguiram mapear a sazonalidade com maior robustez, alcançando um Erro Percentual Absoluto Médio (MAPE) de 4,2%, enquanto os modelos puramente estatísticos apresentaram um MAPE na casa de 6,5%. Tais resultados justificam a necessidade empírica de estabelecer um baseline estatístico (como feito neste projeto) antes de evoluir para algoritmos mais complexos.

### 3.2. Diagnóstico de Estacionariedade (ADF e KPSS)
&emsp;Modelos da família ARIMA exigem que a série temporal seja estacionária (média e variância constantes ao longo do tempo). Para validar isso estatisticamente, aplicamos dois testes complementares: o **Dickey-Fuller Aumentado (ADF)** e o **KPSS**.

1. **Série Original:** O teste ADF não conseguiu rejeitar a hipótese nula ($p-valor = 0.79$) e o KPSS a rejeitou ($p-valor = 0.01$). Isso comprova que a série original **não é estacionária**.
2. **Série Diferenciada:** Aplicamos uma diferenciação de primeira ordem ($d=1$) e uma sazonal ($D=1$). Como resultado, o $p-valor$ do ADF caiu para $4.85 \times 10^{-12}$ e o KPSS não rejeitou sua hipótese nula.

&emsp;Essa prova matemática justifica a escolha dos parâmetros de integração **$d=1$** e **$D=1$** na modelagem do SARIMAX.

````python
def diagnostico_estacionariedade(serie, nome="Série"):
    print(f"=== Diagnóstico de Estacionariedade: {nome} ===")
    adf_res = adfuller(serie.dropna())
    print(f"ADF Stat: {adf_res[0]:.4f} | p-valor: {adf_res[1]:.4e}")
    kpss_res = kpss(serie.dropna(), regression='c', nlags="auto")
    print(f"KPSS Stat: {kpss_res[0]:.4f} | p-valor: {kpss_res[1]:.4f}\n")

# Testa a série original
diagnostico_estacionariedade(df_mensal['PASSAGEIROS_PAGOS_TRATADO'], "Original")

# Testa a série com 1 diferenciação normal (d=1) e 1 sazonal (D=1)
serie_diferenciada = df_mensal['PASSAGEIROS_PAGOS_TRATADO'].diff(1).diff(12).dropna()
diagnostico_estacionariedade(serie_diferenciada, "Diferenciada (d=1, D=1)")
````

&emsp;Os gráficos da série diferenciada sugerem a necessidade de incluir termos autoregressivos (p=1) e de erro de previsão (q=1). A combinação dessas evidências com a sazonalidade justifica a arquitetura *baseline* de **SARIMAX (1, 1, 1) x (1, 1, 1, 12)**.

2. **Tratamento da Anomalia (Variável Exógena):**
   Como o setor aéreo foi severamente impactado entre 2020 e 2021, utilizamos a variável exógena `COVID_DUMMY`. Isso informa ao algoritmo o exato período do choque pandêmico, impedindo que essa anomalia distorça as estimativas de tendência e a sazonalidade natural da série.

3. **Configuração Base do SARIMAX:**
   O modelo foi instanciado com parâmetros iniciais `order=(1, 1, 1)` e `seasonal_order=(1, 1, 1, 12)`. Essa configuração busca capturar a autocorrelação de curto prazo, corrigir tendências lineares e espelhar a forte sazonalidade anual (ciclos de 12 meses) inerente ao turismo e aviação civil.
